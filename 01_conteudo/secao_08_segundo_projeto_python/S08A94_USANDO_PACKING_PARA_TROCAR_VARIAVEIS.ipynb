## Usando Packing Para Trocar Variáveis
***S08 A93***

Dando continuidade a atualização do algoritmo, nessa aulas vamos realizar uma troca de variaveis e seus valores, swap de variaveis eliminando a variavel auxiliar ***proximo***

```python
# Sequência de Fibonacci - v3.0

# 0, 1, 1, 2, 3, 5, 8, 13, 21
# (próximo número é a soma dos dois últimos números)
def fibonacci(limite_final=10000):
    penultimo = 0
    ultimo = 1
    print(f"{penultimo}, {ultimo}", end=", ")
    while ultimo < limite_final:
        penultimo, ultimo = ultimo, penultimo + ultimo
        print(ultimo, end=", ")


if __name__ == "__main__":
    fibonacci(20000)
```
