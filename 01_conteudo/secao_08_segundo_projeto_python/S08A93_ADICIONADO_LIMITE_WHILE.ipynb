## Adicionando Limite Para While
***S08 A93***

Dando continuidade as melhorias existentes para aplicar no fibonacci feito na ultima aula considerando as aplicações, nessa aula vamos colocar um limite no while para ele não executr de forma indefinida,

```python
# Sequência de Fibonacci - v2.0

# 0, 1, 1, 2, 3, 5, 8, 13, 21
# (próximo número é a soma dos dois últimos números)
def fibonacci(limite_final=10000):
    penultimo = 0
    ultimo = 1
    print(f"{penultimo}, {ultimo}", end=", ")
    while ultimo < limite_final:
        proximo = penultimo + ultimo
        print(proximo, end=", ")
        penultimo = ultimo
        ultimo = proximo


if __name__ == "__main__":
    fibonacci(20000)
```
