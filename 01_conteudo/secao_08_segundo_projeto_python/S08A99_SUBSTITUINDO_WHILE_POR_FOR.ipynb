## Substituindo While Por For
***S08 A99***

Nessa aula vamos gerar a sétima e última versão do algoritmo fobonacci que estamos desenvolvendo, basicamento a ultima atualização do fibonacci, realizamos a troca do while pelo for utilizando um range de 0 até a quantidade de números definida dentro dessa sequência fibonacci

```python
# Sequência de Fibonacci - v7.0

# 0, 1, 1, 2, 3, 5, 8, 13, 21
# (próximo número é a soma dos dois últimos números)
def fibonacci(quantidade_numeros=20):
    resultado = [0, 1]
    for _ in range(quantidade_numeros-2): 
        # _ é comumente utilizado na comunidade de python '
        # como varável não utilizada durante a utilização do for
        resultado.append(sum(resultado[-2:]))
    return resultado


if __name__ == "__main__":
    for fib in fibonacci():
        print(fib, end=", ")
```
