## While Com Break
***S08 A98***

Vamos dar continuidade ao processo de autalização do algoritmo de fibonnaci, nesse caso vamos mudar a lógica um pouco e implementar algum recurso, vamos chamar esse algotimos de v6.0, a atualização de implementação desse algoritmo foi exatamente alterar o limite do valor final para a quantidade de numeros que voce deseja na sequencia fibonacci a ser impressa.

```python
# Sequência de Fibonacci - v6.0

# 0, 1, 1, 2, 3, 5, 8, 13, 21
# (próximo número é a soma dos dois últimos números)
def fibonacci(quantidade_numeros=20):
    resultado = [0, 1]
    while True:
        resultado.append(sum(resultado[-2:]))
        if quantidade_numeros == len(resultado):
            break
    return resultado


if __name__ == "__main__":
    for fib in fibonacci():
        print(fib, end=", ")
```
