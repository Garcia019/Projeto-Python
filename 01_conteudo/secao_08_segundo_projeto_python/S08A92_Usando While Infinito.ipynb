 Na seção 8 vamos dar inicio a processo de aprendizado conforme fizemos na secão 06 onde cada aula era uma melhoria aplicada ao algoritmo em desenvolvimento, nesse caso vamos desenvolver um algoritmo para calcular a fibonacci, considerando o uso das estruturas de controle nesse exemplo falado que seria a estrutura de fibonacci, vamos usar o arquivo fibonacci_vx.py para alterar a cada aula

```python
# Sequência de Fibonacci - v1.0

# 0, 1, 1, 2, 3, 5, 8, 13, 21
# (próximo número é a soma dos dois últimos números)
def fibonacci():
    penultimo = 0
    ultimo = 1
    print(f"{penultimo}, {ultimo}", end=",")
    while True:
        proximo = penultimo + ultimo
        print(proximo, end=",")
        penultimo = ultimo
        ultimo = proximo


if __name__ == "__main__":
    fibonacci()
```
