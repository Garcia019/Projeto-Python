## Recursividade

A partir dessa aula vamos trasnformar a versão fibonacci desenvolvida nas últimas aulas para uma versão recursiva, basicamento recursividade é utilizar um método para chamar ele mesmo dentro do python porem com limites definidos para finalizar a recursividade, é essencial que exista uma condição de parada para que não se torne um loop infinito e de erro.

```python
#Exemplo de Recursividade

def imprimir(maximo, atual):
    if atual >= maximo: #condição de parada
        return # Sai do método sem retornar nada 
    print(atual,end=", ")
    imprimir(maximo, atual + 1) # utiliza recursividade garantido que teha um condição que para a recursividade nesse caso é quando o atual é maior 
    # ou igual ao maximo


imprimir(10, 0)
```
