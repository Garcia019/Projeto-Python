Vamos dar continuidade no algoritmo de recursividade de fibonacci, versão 2.0 e última conforme orientações do professor,

```python
# Sequência de Fibonacci Recursiva - v2.0

# 0, 1, 1, 2, 3, 5, 8, 13, 21
# (próximo número é a soma dos dois últimos números)
def fibonacci(resultado=(0, 1), quantidade_numeros=20):
    if len(resultado) == quantidade_numeros: # condição de parada dentro da recursividade
        return resultado
    return fibonacci((resultado + (sum(resultado[-2:]), )) , quantidade_numeros)


if __name__ == "__main__":
    for fib in fibonacci():
        print(fib, end=", ")
```
