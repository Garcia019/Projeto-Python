Nessa lição vamos restruturar a função fibonacci para utilizar a recursividade como padrão, no caso tentei iniciar a implementação com lista porem deu errado por conta do endereço de memoria responsavel pela lista ser o mesmo todas as vezes que voce chamar a mesma função

```python
# Sequência de Fibonacci Recursiva - v1.0

# 0, 1, 1, 2, 3, 5, 8, 13, 21
# (próximo número é a soma dos dois últimos números)
def fibonacci(resultado=(0, 1), quantidade_numeros=20):
    return resultado if len(resultado) == quantidade_numeros else \
    fibonacci((resultado + (sum(resultado[-2:]), )) , quantidade_numeros)


if __name__ == "__main__":
    for fib in fibonacci():
        print(fib, end=", ")
```
