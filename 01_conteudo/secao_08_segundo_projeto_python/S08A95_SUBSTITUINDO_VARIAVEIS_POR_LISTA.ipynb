## Substituindo Variáveis Por Lista
***S08 A95***

Dando continuidade a ciclo de atualizações do algorimos de fibonacci, nessa aula vamos substituir as variaveis por uma lista e trabalhar com seus indices, removemos os prints() de dentro das funções e retornamos uma lista com a sequência fibonacci desejada

```python
# Sequência de Fibonacci - v4.0

# 0, 1, 1, 2, 3, 5, 8, 13, 21
# (próximo número é a soma dos dois últimos números)
def fibonacci(limite_final=10000):
    resultado = [0, 1]
    while resultado[-1] < limite_final:
        resultado.append(resultado[-1] + resultado[-2])
    return resultado


if __name__ == "__main__":
    for fib in fibonacci(20000):
        print(fib, end=", ")
```
