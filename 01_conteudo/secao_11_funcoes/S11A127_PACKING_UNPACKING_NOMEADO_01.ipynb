Nessa aula vamos testar e gerar exemplos do Packing & Unpacking nomeado **kwargs que empacota e desempacota um dicionário, com aplicações reais

```python
# Exemplos Packing_Nomeado
# **kwargs

# Packing
def resultado_f1(**podium):
    # Nesse caso ao utilizar o **podium ele vai pegar todos os parametros passados e vai empacotar em um variavel do tipo dicionario
    # A partir disso conseguimos percorrer essa variável como dicionário 
    for posicao, piloto in podium.items():
        print(f"{posicao} -> {piloto}")

# Unpacking
def resultado_f1_packing(primeiro, segundo, terceiro):
    # Nesse caso recebos 3 parâmetros para imprimir na posição definida
    print(f"1º - {primeiro}")
    print(f"2º - {segundo}")
    print(f"3º - {terceiro}")


if __name__ == "__main__":
    # nesse caso passando esse parametros para ele, a função empacota essas variáveis em um dicionário só
    resultado_f1(primeiro="L. Hamilton",
              segundo="M. Verstappen",
              terceiro="S. Vettel")

    # Aqui criamos um variável do tipo dicionário com o podium e desempacotamos ao chamar a função entregando separadamente as variáveis
    # No caso ele chama parâmetros da o nome dos variáveis/parâmetros as keys no dicionario e o conteudo seria o valor da chave
    podium = {"primeiro": "L. Hamilton",
              "segundo": "M. Verstappen",
              "terceiro": "S. Vettel"}

    resultado_f1_packing(**podium,)
```
