Nessa aula vamos passar um pouco sobre o unpacking, vamos introduzir um conceito novo no aprendizado para depois aplicar no gerador HTML,

```python
# Exemplos Packing & Unpacking de listas e tuplas utilizando em funções

def soma_2(a,b):
    return a + b

def soma_3(a, b, c):
    return a + b + c

# Essa função foi criada com a intensão de somar n valores
# Utilizando o parâmetro *numeros você transforma os parâmetros passados em um tupla
# o caractere "*" ele faz o "empacotamento do parametros independente de quantos são e transforma-os em um tupla"
def soma_n(*numeros):
    soma = 0
    for n in numeros:
        soma += n
    return soma

if __name__ == "__main__":
    print(soma_2(2, 3))
    print(soma_3(2, 4, 6))

    # packing
    print(soma_n(1)) # tupla de 1 valor 
    print(soma_n(1, 1)) # tupla de 2 valores
    print(soma_n(1, 1, 1, 1, 1, 1, 1)) # tupla de 7 valores

    # unpacking
    nums = (1, 2, 3)
    print(soma_n(*nums))
    # nessa caso ao utilizar o caractere "*" na chamada da função utilizando a varaviavel "nums" que reprensenta uma tupla
    # o caractere "*" é responsável por "desempacotar" a tupla e ao inves de enviar apenas um tupla ou lista
    # ele envia tres parametros separados para a função conforme o exemplo, porem podem ser n paramentros proporcional ao tamanho da lista ou tupla
    
```
