Nessa seção vamos trabalhar um assunto muito importante para programação que são as funções, uma função é um sequencia de passos que processa ou realiza uma tarefa com base em dados(entradas) ou não, nesse caso um função é também um algoritmo que pode ser reutilizado sempre que necessário utilizando seu nome ou referencia, recebendo argumentos ou não, função em python é um objeto, e um objeto pode se comportar como função, em uma função temos dois tipos de parametros, Posicional e Nomeado,

Parâmetro Posicional, quando definimos uma função com p1, p2, p3 sendo p de parametro, ao chamar essa mesma função passando os parametros(2, 3, 4) os argumentos seram mapeados em ordem em relação aos parametros definidos por você na criação da funcão, posicionados em relação ao que foi passado

Já o parâmetro nomeado, conseguimos definir qual dos argumentos seram utilizados para qual variavel o executor escolhe, qual argumentos serao utilizados por qual variável

Outros dois tipos de parametros muito famosos são *args e **kwargs , a importancia desses parametros são os ** que no caso no parametro *args gera uma tupla e no **kwargs gera um dicionário

O python permite retornar apenas um resultado dentro da função porem esse retorno pode ser todo tipo de dado existente desde que faca sentido ao uso da função, 

```python
# Exemplo de Parametros Posicional
def funcao(p1, p2, p3):
    print(p1, p2, p3)

funcao(2, 3, 4) # Seguindo a lógica do algoritmo o número 2 será o p1, o número 3 será o p2 e o número 4 será o p3

#Exemplo de Parametro Nomeado

funcao(2, p3=1, p2=2) 
# Nesse exemplos podemos perceber que o executor escolhe os destinos extados de cada parâmetro não importando a ordem dos definidos
```
