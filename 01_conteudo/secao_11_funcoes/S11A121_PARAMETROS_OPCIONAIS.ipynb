Dando conmtinuidade ao estudo das funções nessa aula agora vamos realizar exemplos utilizando a metodologia de parâmetros opcionais para entender e aplicar mesmo que ja utilizada em algum exemplo anterior durante o curso.

```python
# Gerador HTML - version-v1.0

def tag_bloco(texto, classe="success"): # classe é um parametro opcional nesse algoritmo
    return f"<div class='{classe}'>{texto}</div>"

if __name__ == "__main__":
    # Testes (assertions)
    # Assertions é utilizado para testar funcões ou codigos para verificar sua aplicabilidade conforme solicitado
    assert tag_bloco("Incluído Com Sucesso!") == \
        "<div class='success'>Incluído Com Sucesso!</div>"
    assert tag_bloco("Impossível excluir!", "error") == \
        "<div class='error'>Impossível excluir!</div>"
    print(tag_bloco("bloco"))
```
