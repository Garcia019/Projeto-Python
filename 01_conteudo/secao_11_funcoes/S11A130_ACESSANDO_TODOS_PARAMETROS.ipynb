## Acessando Todos os Parâmetros
***S11 A130***

Nessa aula vamos gerar uma função que consegue capturar todos os parâmetros passados de uma forma genérica,

```python
# Todos os Parâmetros Funções
# *args representa todos os parâmetros posicionaios
# **kwargs representa todos os parâmetros nomeados


def todos_params(*args, **kwargs):
    print(f"args: {args}")
    print(f"kwargs: {kwargs}")


if __name__ == "__main__":
    todos_params("a", "b", "c")
    # nesse caso os parâmetros posicionais 1,2,3 são empacotados pelo *args
    # e os parâmetros nomeados são empacotados legal=True e valor=12.99 pelos *kwargs
    todos_params(1, 2, 3, legal=True, valor=12.99)
    # Separa adequadamente independente da ordem
    todos_params("Ana", False, [1, 2, 3], tamanho="M", fragil=False)
    # todos_params(primeiro="João", "Maria")
    # Nesse caso a linha de código acima gera um erro 
    # pois o primeiro parametro passado a função é um parâmetro de *args
    # e o último é de **kwargs, ou seja, um argumento posicional
    # esta localizado após um argumento nomeado
    
```
