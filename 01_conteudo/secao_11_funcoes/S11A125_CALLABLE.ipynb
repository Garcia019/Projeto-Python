Nessa aula vamos dar uma olhada rapida na programação funcional que sera um pouco das aulas dos últimos arquivos, porém vamos adiantar nessa aula um conceito na programação funcional callable para evoluir o algoritmo

```python
# Programação Funcional - Conceito Callable

# criando um função que o papel dela é invocar outra função

def executar(funcao):
    if callable(funcao): # Retorna True caso o parametro funcao seja um callable (objeto que pode ser chamado)
        funcao()


def bom_dia():
    print("Bom dia!")


def boa_tarde():
    print("Boa Tarde!")


if __name__ == "__main__":
    executar(bom_dia)
    executar(boa_tarde)
    executar(1) # Objeto int "1" não é um callable passivel de ser chamado
```
