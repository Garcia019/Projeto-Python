## Generators com FOR
***S010 A117***

Nesse exemplos vamos dar continuidade aos estudo de Generators, porém aplicando ao uso do for e mostrando um pouco da função next() encapsulada diretamente no FOR

```python
# Generator - version-2.0
# (expressão for item in list if condicional)

generator = (numero ** 2 for numero in range(10) if numero % 2 ==0)
for numero in generator: # O for reconhece um generator e ja encapsula um next() oara geração de cada arquivo por linha
    print(numero)
```
