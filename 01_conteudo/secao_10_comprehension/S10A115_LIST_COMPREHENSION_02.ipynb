Dando continuidade aos exemplos do list comprehension e sua aplicabilidade em python, nesse caso vamos criar uma lista em python utilizando o list comprehension porem utilizando um condicional como base.

```python
# Comprehension - version-2.0
# [expressão for item in list if condicional]

dobro_dos_pares = [numero * 2 for numero in range(10) if numero % 2 == 0]
# Nesse caso eu gero uma lista utilizando o for de numeros de 0 ao 9
# Conforme o Range determinado, além de pegar somente o dobro pares pela condição mostrada
print(dobro_dos_pares)

# Criando a mesma lista sem utilizar o list comprehension

dobro_dos_pares_natual = []
for numero in range(10):
    if numero % 2 ==0:
        dobro_dos_pares_natual.append(numero*2)
print(dobro_dos_pares_natual)
```
