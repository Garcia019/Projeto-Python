Nessa aula vamos dar continuidade ao aprendizado do comprehension, porem realizando algumas alterações na sintaxe do codigo e entendendo as mudanças realizadas,

```python
# Generator - version-1.0
# (expressão for item in list if condicional)

import sys

# gerador de número que gera automaticamente o proximo numero na logica determinada através do ()
# consome menos memória que o list comprehension que no caso gera a lista completa
# nesse caso o generator gera os dados de forma continua mas aous poucos
# e nao todos os dados de uma vez como no list comprehension
generator = (numero ** 2 for numero in range(1000) if numero % 2 ==0) 
list_comprehension = [numero ** 2 for numero in range(1000) if numero % 2 ==0]

print(next(generator)) # Saida 0
print(next(generator)) # Saida 4
print(next(generator)) # Saida 16
print(next(generator)) # Saida 36
print(next(generator)) # Saida 64
# print(next(generator)) # Saida error!

# Isso mostra que o generator tem tamanho fixo e quando tratamos de valores muitos grandes
# É melhor para otimizar o uso de memoria da maquina
# Esses prints mostram a diferença de valor para valores grandes como 1000 numeros
print("O tamanho de memoria utilizado no método Generator é: ", sys.getsizeof(generator))
print("O tamanho da memoria utilizado no método List Comprehension é: ", sys.getsizeof(list_comprehension))




```
