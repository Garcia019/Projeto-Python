Dando inicio ao capítulo/seção 10 o list comprehension é uma forma mais consiza para criar lista, sintaxe especifica do python para criar lista muito rapido a partir de uma unica linha, vamos realizar alguns exemplos,

```python
# Comprehension - version-1.0
# [expressão for item in list]

dobros = [i * 2 for i in range(10)] # isso cria uma lista com o dobro de cada i que é rodado no for, no caso de 0 a 9
print(dobros)

# Criando a mesma lista sem utilizar o list comprehension

dobros_natural = []
for numeros in range(10):\
    dobros_natural.append(numeros * 2)
print(dobros_natural)
```
