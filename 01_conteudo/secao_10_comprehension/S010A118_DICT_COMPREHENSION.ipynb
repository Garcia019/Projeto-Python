## Dict Comprehension
***S010 A118***

Nessa aula vamos dar seguimento ao aprendizado do comprehension porem nesse caso o uso com dicionários,

```python
# Dict Comprehension - version-1.0
# {chave: valor expressão for item in list if condicional}

dicionario = {numero: numero * 2 for numero in range(10) if numero % 2 == 0}
print(dicionario)

for numero, dobro in dicionario.items():
    print(f"{numero} x 2 = {dobro}")
```
