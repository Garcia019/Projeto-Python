Nessa aula vamos dar continuidade na forma de otimizar a abertura de arquivos com novas atribuições corrigindo o problema relatado na aula passada, estudando um pouco sobre o métrodo .strip()


```python
# Método Strip Exemplos Práticos

frase1 = "   teste abc   "
frase2 = "   teste   abc   "
frase3 = "00000 teste abc 00000"

print("Frase 1: '{}' - SEM STRIP()".format(frase1))
print("Frase 1: '{}' - COM STRIP()".format(frase1.strip()))# remove os espaços exesivos que envolvem a string

print("Frase 2: '{}' - SEM STRIP()".format(frase2))
print("Frase 2: '{}' - COM STRIP()".format(frase2.strip())) # remove os espaços exesivos que envolvem a string porem não os do meio

print("Frase 3: '{}' - SEM STRIP()".format(frase3))
print("Frase 3: '{}' - COM STRIP('0')".format(frase3.strip("0"))) # remove os "00000" porem mantem os espacos entre os zeros e a string interna

print("Frase 3: '{}' - SEM STRIP()".format(frase3))
print("Frase 3: '{}' - COM STRIP('0').split()".format(frase3.strip("0").strip())) # remove os "00000" e os espacos entre os zeros e a string interna

# Método Strip remove com certeza toda quantidade repetitiva que envolve uma string ou dado como padrão o " ", no caso espaços em brancos
# Remover também quebra de linhas

```

```python
# IO - version_3.0

arquivo = open("pessoas.csv")

for registro in arquivo:
    print("Nome: {}, Idade: {}".format(*registro.strip().split(",")))

arquivo.close()
```
