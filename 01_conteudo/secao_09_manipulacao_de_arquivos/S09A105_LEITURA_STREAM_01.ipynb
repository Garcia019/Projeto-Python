Dando continuidade as aulas de manipulação de arquivos vamos gerar a segunda versão da manipulação de arquivos, utilizando como exemplo a abertura do arquivo pessoas.csv, nessa versão vamos utilizar o método stream, essa verão seria mais simples e mais otimizada na forma como os recursos são consumidos.

A diferença entre o primeito algoritmo e esse segunda é que eu pego o arquivo leio todo o arquivo com ***dados = arquivo.read()***, armazenando todo o tamanho do arquivo na memória do computador durante a execução, ou seja, durante a execução se o arquivo tiver 100MB de memoria ele será carregado no computador 100MB de memória, ja no segundo caso você pega o arquivo não fecha ele e le o aquivo sobre demanda, por partes, e isso faz uma diferença absurda, pois você apenas carrega parte do sistema sem carregar o todo, assim como um vídeo do youtube, método streaming.

```python
# IO - version_2.0
# Abrindo arquivo utilizando open()
arquivo = open("pessoas.csv")

# Percorrendo o arquivo de início sem armazena-lo

for registro in arquivo:
    # Processando os dados de forma rápida e automática
    # Gerando um linha a mais na impressão pois o \n está associado
    print("Nome: {}, Idade: {}".format(*registro.split(",")))

# Logo após fechando o arquivo depois de ter processado
arquivo.close()

"""
CÓDIGO ELIMINADO NO PROCESSO DE MELHORIA
dados = arquivo.read()

arquivo.close()

for registro in dados.splitlines():
    print("Nome: {}, Idade: {}".format(*registro.split(",")))

"""
```
