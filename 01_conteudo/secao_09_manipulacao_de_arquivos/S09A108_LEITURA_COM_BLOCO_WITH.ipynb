Dendo continuidade ao versionamento das versões e agora na versão 5 vamos simplificar o último algoritmo v4.0, com isso vamos realizar a implementação do ***with open("pessoas.csv") as arquivo:*** nesse caso isso cria um bloco onde a abertura do arquivo é identificado como a variavel arquivo, e disponibiliza o arquivo aberto apenas dentro do bloco gerado através do comando mostrado acima, com isso  é garantido que o arquivo sera fechado quando todas as linhas de código dentro do bloco foram executadas, ou seja, é garantido que o recurso utilizado de memoria para abertura do arquivo no programa sera redistribuido.

Esse bloco tem a mesma funcão o bloco try: finally da v4.0 porem de forma mais simples de executar e implementar.

```python
# IO - version_5.0

with open("pessoas.csv") as arquivo:
    for registro in arquivo:
        # Removi o "*" para gerar um erro e mostrar a execução do finally
        print("Nome: {}, Idade: {}".format(*registro.strip().split(",")))

if arquivo.closed:
    print("Arquivo já foi fechado!")

```
