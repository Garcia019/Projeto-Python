Nessa aula vamos utilizar um módulo disponível na API do python chamado CSV, vamos fazer a leitura do arquivo io_csv, com o auxilio desse modulo csv do pyhthon para entender e saber como funciona e como utilizar corretamente,


Basicamente o módulo csv ele facilita a forma de interagir com o arquivo csv com alguns funções de tratamentos de arquivo csv ja prontas, basicamente ele ja trata os ***"\n" - Enter, as ","*** todos esses detalhes e te entrega um arquivo limpo, no caso a função ***"csv.reader()"*** ele ja pega o csv aberto e realizar todo a limpeza e te entrega uma lista com os arquivos separados em ordem.


Como se ***"csv.reader(registro)"*** substituísse ***registro.strip().split(",")***, porem obviamente apenas funciona com csv pois ele trata o arquivo como um csv padrão

```python
# IO_SCV - version 1.0

import csv

with open("pessoas.csv") as entrada:
    for pessoa in csv.reader(entrada):
        print("Nome: {}, Idade: {}".format(*pessoa))
```
