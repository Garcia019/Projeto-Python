Nessa aula após ter criado manualmente o arquivo, pessoa.csv vamos desenvolver um algoritmo que faça a leitura básica do arquivo definido, é necessário que após abrir o arquivo em python seja realizado o fechamento do arquivo de forma explicita ou implicita para ter um cotrole melhor da quantidade de memoria alocada e utilizada pela aplicação

```python
# IO - version_1.0
# Abrindo arquivo utilizando open()
arquivo = open("pessoas.csv")

# Lendo o arquivo aberto
dados = arquivo.read()

# Fechando arquivo após capturar os dados .close()
arquivo.close()

# .splitlines() é um método para separar cada linha do arquivo
# Considerando o parâmetro padrão \n entre outros padrões utilizados para separar linha
for registro in dados.splitlines():
    # O registro em si ele esta apresentado como "Maria,45"
    # Utilizando o split conseguimos criar uma lista
    # [Maria, 45] e utilizar da forma que precisamos nesse caso
    # Reposicionando utilizando .format(*)
    print("Nome: {}, Idade: {}".format(*registro.split(",")))
```
