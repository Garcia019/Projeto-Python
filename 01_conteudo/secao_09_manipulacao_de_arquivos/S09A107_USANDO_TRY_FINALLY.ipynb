## Usando Try Finally
***S09 A107***

Nessa aula vamos dar continuidade na alteração do algoritmo considerando o uso do try finally, implementendo essa funcionalidade no algoritmo de versão 4, try finnaly basicamente é realizar alguma ação mesmo que o algoritmo pertencente ao bloco try de algum problema ou erro durante a execução ao inves de parar toda a aplicação ou algoritmo você não executa função tentada mas executa todo o resto vamos aplicar o exemplo abaixo.

Utilizamos considerando que o trecho do código é arriscado para não prejudicar outras partes do código deixamos como try: finally:, o finally sempre sera executado ou não independente do try:, porem garantimos só a execução do bloco finally não qualquer outro código após ele 

```python
# IO - version_4.0

try:

    arquivo = open("pessoas.csv")

    for registro in arquivo:
        # Removi o "*" para gerar um erro e mostrar a execução do finally
        print("Nome: {}, Idade: {}".format(*registro.strip().split(",")))

"""
Utilizando except IndexError que é exatamente
O erro que aparece ao remover o "*" e nesse caso
com essa funcionalidade o código continua seu
caminho definido até encontrar ou erro 
"""

except IndexError:
    pass
finally:
    print("Finally")
    arquivo.close()

if arquivo.closed:
    print("Arquivo já foi fechado!")
```
