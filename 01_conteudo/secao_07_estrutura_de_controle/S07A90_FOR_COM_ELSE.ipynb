## FOR Com ELSE
***S07 A90***

Nessa aula vamos dar continuidade dno exemplo da última aula agora aplicando o uso do ELSE atrelado a um for dependendo da situação que foi aplicada anteriomente,

```python
PALAVRAS_PROIBIDAS = ("futebol", "religião", "política")
textos = [
    "João gosta de futebol e política",
    "A praia foi divertida",
]

for posicao, texto in enumerate(textos):
    for palavra in texto.lower().split():
        if palavra in PALAVRAS_PROIBIDAS:
            print(f"Texto número {posicao + 1} possui pelo menos um palavra probida: {palavra}")
            break
    else:
        print(f"Texto número {posicao + 1} autorizado: {texto}!")
```
