## FOR Sem ELSE

Nessa aula vamos mostrar exemplos da implementação do FOR com else e sem else, como isso vamos repassar algumas situações e utilizar como base as últimas aulas.  

Outro detalhe abordado nessa aula é sobre as constantes declaradas no python não necessariamente voce tem essa funcionalidade de criara um variavel imutavel (constante) como em outras linguagens como javascript porem como forma de notacao foi adotado que durante a contrução de um algoritmos todas as variaveis consideradas constantes devem estar em letras maisculas para identifica-las melhor.



```python
PALAVRAS_PROIBIDAS = ("futebol", "religião", "política")
textos = [
    "João gosta de futebol e política",
    "A praia foi divertida",
]

for posicao, texto in enumerate(textos):
    found = False
    for palavra in texto.lower().split():
        if palavra in PALAVRAS_PROIBIDAS:
            print(f"Text número {posicao + 1} possui pelo menos um palavra probida: {palavra}")
            found = True
            break
    if not found:
        print(f"Texto número {posicao + 1} autorizado: {texto}!")
```
