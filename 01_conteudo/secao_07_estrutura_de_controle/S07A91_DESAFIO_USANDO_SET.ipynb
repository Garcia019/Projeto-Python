Com o desafio realizado tivemos alguns insights do professor em relação otimizar o algoritmo por exemplo, uma lista vazia em python [] é considerada False quando utilizada em uma estrutura condicional, bem como um dicionário vázio, uma tupla ou até mesmo um conjunto(SET), com isso a versão 1.0 foi feito por mim, e versão 2.0 com algumas melhorias de sintaxe porem os dois programas funcionam adequadamente

```python
# Desafio Utilizando SET - version 1.0

PALAVRAS_PROIBIDAS = {"futebol", "religião", "política"}
frases = [
    "João gosta de futebol e política",
    "A praia foi divertida",
]

for indice, frase in enumerate(frases):
    frase_natural = frase.lower().split()
    conjunto_palavras = set(frase_natural)
    if len(PALAVRAS_PROIBIDAS.intersection(conjunto_palavras)) != 0:
        print(f"""Texto número {indice + 1} possui palavra probida:
{PALAVRAS_PROIBIDAS.intersection(conjunto_palavras)}""")
    else:
        print(f"Texto número {indice + 1} autorizado: {frase}!")
```

```python
# Desafio Utilizando SET - version 2.0

PALAVRAS_PROIBIDAS = {"futebol", "religião", "política"}
frases = [
    "João gosta de futebol e política",
    "A praia foi divertida",
]

for indice, frase in enumerate(frases):
    intersecao = PALAVRAS_PROIBIDAS.intersection(set(frase.lower().split()))
    if intersecao:
        print(f"""Texto número {indice + 1} possui palavra probida: {intersecao}""")
    else:
        print(f"Texto número {indice + 1} autorizado: {frase}!")
```
