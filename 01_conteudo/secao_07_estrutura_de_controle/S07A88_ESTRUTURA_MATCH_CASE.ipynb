Essa aula foi criada após o lançamenmto do curso, pois durante as últimas aulas foram mostradas formas de simular a estrutura SWITCH, porém a partir da versão 3.10 do python foram implementada uma nova estrutura de controle que tem o mesmo comportamente da estrutura SWITCH, ESSA ESTRTURA FOI CHAMADA DE ***MATCH CASE***

Através disso vamos refazer os exercício realizados nas últimas duas aulas considerando essa nova estrutura como forma de praticar, porem vamos mostrar a sintaxe básica dessa estrutura.

```python
# Sintaxe básica da estrutura match case

valor = int(input("Digite um número:"))

match valor:
    case 1:
        print("Código para caso 1!")
    case 2:
        print("Código para caso 2!")
    case 3:
        print("Código para caso 3!")
    case 4:
        print("Código para caso 4!")
    case 5:
        print("Código para caso 5!")
    case _:
        print("Código para qualquer outro caso não definido")
```

```python
# Desafio Função Grupo Dia da Semana - Utilizando o match case

def grupo_semana(dia):
    match dia:
        case 1 | 7:
            return "Final de Semana"
        case 2 | 3 | 4 | 5 | 6:
            return "Dias úteis"
        case _:
            return "Grupo Inválido"

def get_dia_semana(dia):
    dias = {
        1: "Domingo",
        2: "Segunda",
        3: "Terça",
        4: "Quarta",
        5: "Quinta",
        6: "Sexta",
        7: "Sábado"
    }
    return dias.get(dia, "Dia Inválido")

    
if __name__ == "__main__":
    for dia in range(0, 9):
        print(f"{dia} é {get_dia_semana(dia)} e pertence ao {grupo_semana(dia)}!")
```
